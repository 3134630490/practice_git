```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv('./deepseek_data.csv')
```

#### 1.数据加载与探索

```python
print("\n前五行数据为：")
display(data.head())

print("\n数据信息：")
print(data.info())

print("\n数据统计描述：")
print(data.describe())

print("\n数据类型为：")
print(data.dtypes)

print("\n数据形状：")
print(f'行数：{data.shape[0]},列数：{data.shape[1]}')
```

#### 2.数据清洗

```python
data['order_date'] = pd.to_datetime(data['order_date'])

# 检查缺失值
print('\n缺失值检查')
print(data.isnull().sum())

#处理缺失值
data['remarks'] = data['remarks'].fillna('无备注')

#处理重复值
print('\n重复值数量')
print(f"\n=== 重复订单数: {data.duplicated('order_id').sum()} ===")
```
